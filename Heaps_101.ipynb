```python
# Implementing Min Heap and Heapify an array
# TC : O(n), SC : O(1)

A = [-4, 1, 5, 3, 10, 9, -2, 6]

import heapq
heapq.heapify(A)

A
```

```python
# Implementing Max Heap

B = [-4, 1, 5, 3, 10, 9, -2, 6]
n = len(B)

for i in range(n):
  B[i] = -B[i]

heapq.heapify(B)

print(B, "\n")

max_heap = -heapq.heappop(B)
print(max_heap)
```

```python
# Heap Push (Inserting an element)
# TC : O(log n)

heapq.heappush(A, 4)

A
```

```python
# Heap Pop (Deleting the min element)
# TC : O(log n)

minn = heapq.heappop(A)

A, minn
```

```python
# Heap Sort
# TC : O(n logn), SC : O(n)... coz we are creating a new list

def heapSort(arr):
  heapq.heapify(arr)
  n = len(arr)
  new_list = [0] * n

  for i in range(n):
    minn = heapq.heappop(arr)
    new_list[i] = minn

  return new_list

heapSort([-4, 1, 5, 3, 10, 9, -2, 6])
```
