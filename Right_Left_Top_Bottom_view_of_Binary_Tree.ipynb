# TreeNode Definition

```python
from collections import deque

class TreeNode:
  def __init__(self, val = 0, left = 0, right = 0):
    self.val = val
    self.left = left
    self.right = right


def arr_to_tree(arr):

  if not arr or arr[0] is None:
    return None

  root = TreeNode(arr[0])
  queue = deque([root])

  i = 1

  while queue and i < len(arr):
    current = queue.popleft()

    if i < len(arr) and arr[i] is not None:
      current.left = TreeNode(arr[i])
      queue.append(current.left)
    i += 1

    if i < len(arr) and arr[i] is not None:
      current.right = TreeNode(arr[i])
      queue.append(TreeNode(arr[i]))
    i += 1

  return root
```

```python
class Solution:
  ''' ------------------------------------------------------------
  ---------------- Right View -------------------------------------
  -----------------------------------------------------------------'''
  def rightsideView(self, root):
    res = []
    self.rightView(root, 0, res)
    return res

  def rightView(self, node, level, res):
    if not node:
      return
    if len(res) == level:
      res.append(node.val)

    self.rightView(node.right, level + 1, res)
    self.rightView(node.left, level + 1, res)

    ''' ------------------------------------------------------------
  ---------------- Left View -------------------------------------
  -----------------------------------------------------------------'''
  def leftsideView(self, root):
    res= []
    self.leftView(root, 0, res)
    return res

  def leftView(self, node, level, res):
    if not node:
      return
    if len(res) == level:
      res.append(node.val)

    self.leftView(node.left, level + 1, res)
    self.leftView(node.right, level + 1, res)
```

```python
arr = [1, 2, 3, None, 5, None, 4]

root = arr_to_tree(arr)

sol = Solution()
rightView = sol.rightsideView(root)
leftView = sol.leftsideView(root)

print("Right View :",rightView)
print("\nLeft View :",leftView)
```
