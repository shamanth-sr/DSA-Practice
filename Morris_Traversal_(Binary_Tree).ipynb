```python
class TreeNode:
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

arr = [1, 2, 3, 4, 5, None, None, None, None, None, 6]

nodes = [TreeNode(val) if val is not None else None for val in arr]

child_idx = 1
for current_node in nodes:
  if current_node is None:
    continue

  if child_idx < len(nodes):
    current_node.left = nodes[child_idx]
    child_idx += 1

  if child_idx < len(nodes):
    current_node.right = nodes[child_idx]
    child_idx += 1

root = nodes[0]
```

```python
'''---------------------------------------------------------------
-----------------------InOrder Traversal---------------------------
--------------------------------------------------------------------'''

def inorderTraversal(root):
  result = []
  curr = root
  while curr:

    # First case - left is not exist
    if curr.left is None:
      result.append(curr.val)
      curr= curr.right

    # Second case - left it exists
    else:
      prev = curr.left
      while prev.right and prev.right != curr:
        prev = prev.right

      # Making the thread to the curr
      if prev.right is None:
        prev.right = curr
        curr = curr.left

      # already thread exists
      else:
        prev.right = None
        result.append(curr.val)
        curr = curr.right
  return result


```

```python
'''---------------------------------------------------------------
-----------------------PreOrder Traversal---------------------------
--------------------------------------------------------------------'''
def preOrderTraveral(root):
  result = []
  curr = root
  while curr:

    # Case 1 : Left node is not present
    if curr.left == None:
      result.append(curr.val)
      curr= curr.right

    # Case 2: Left node is present
    else:
      prev = curr.left
      while prev.right and prev.right != curr:
        prev = prev.right

      if prev.right == None:
        prev.right = curr
        result.append(curr.val)
        curr = curr.left

      if prev.right == curr:
        prev.right == None
        curr = curr.right

  return result
```

```python
print(inorderTraversal(root))
print(preOrderTraveral(root))

```
